```python
from langchain_text_splitters import HTMLHeaderTextSplitter


html_string = """
<!DOCTYPE html>
<html>
<body>
    <div>
        <h1>Respiratory System</h1>
        <p>The respiratory system is responsible for taking in oxygen and expelling carbon dioxide.</p>
        
        <div>
            <h2>Left Lung</h2>
            <p>The left lung is slightly smaller than the right lung to accommodate the heart.</p>
            <h3>Left Superior Lobe</h3>
            <p>This upper section of the left lung brings in air through the left main bronchus.</p>
            <h3>Left Inferior Lobe</h3>
            <p>The lower section rests comfortably on the diaphragm muscle.</p>
        </div>
        
        <div>
            <h2>Right Lung</h2>
            <p>The right lung is shorter but wider than the left lung, divided into three distinct parts.</p>
            <h3>Right Superior Lobe</h3>
            <p>The highest lobe of the right side, located near the collarbone.</p>
            <h3>Right Middle Lobe</h3>
            <p>The smallest lobe, separated by the horizontal and oblique fissures.</p>
            <h3>Right Inferior Lobe</h3>
            <p>The largest lobe on the right side, sitting directly above the liver.</p>
        </div>
    </div>
</body>
</html>
"""


headers_to_split_on=[
    ("h1", "Header 1"),
    ("h2", "Header 2"),
    ("h3", "Header 3"), 
    ]
html_splitter = HTMLHeaderTextSplitter(headers_to_split_on)
html_header_splits=html_splitter.split_text(html_string)
html_header_splits
```

```python
url ="https://www.britannica.com/biography/Shivaji"

headers_to_split_on=[
    ("h1", "Header 1"),
    ("h2", "Header 2"),
    ("h3", "Header 3"),
    ("h4", "Header 4"),
    ("h5", "Header 5"),
    ]
html_splitter = HTMLHeaderTextSplitter(headers_to_split_on)
html_header_splits= html_splitter.split_text_from_url(url)
html_header_splits
```
