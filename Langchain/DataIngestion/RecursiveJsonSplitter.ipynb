```python
import json
import requests

json_data=requests.get("https://api.github.com/repos/psf/requests").json()
json_data   
```

```python
from langchain_text_splitters import RecursiveJsonSplitter
json_splitter=RecursiveJsonSplitter(max_chunk_size=300)
json_chunks=json_splitter.split_json(json_data)
json_chunks
```

```python
for chunk in json_chunks[:3]:
    print(chunk)
```

# THe Splitter can also output doc

```python
docs=json_splitter.create_documents(json_chunks)
for doc in docs[:3]:
    print(doc)
```

```python
texts=json_splitter.split_text(json_data)
print(texts[0])
print(texts[1])
```
