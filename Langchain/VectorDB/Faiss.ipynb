```python
from langchain_community.document_loaders import TextLoader
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import OllamaEmbeddings
from langchain_text_splitters import CharacterTextSplitter

loader=TextLoader("speech.txt")
documents=loader.load()
text_splitter=CharacterTextSplitter(chunk_size=200, chunk_overlap=30)
docs=text_splitter.split_documents(documents)
```

```python
docs
```

```python
embeddings = OllamaEmbeddings(model="nomic-embed-text")


db=FAISS.from_documents(docs, embeddings)
db
```

```python
#Querying the vector store
query = "What is the main topic of the speech?"
docs = db.similarity_search(query)
docs[0].page_content
```

```python
#As. a. Retrieval

retriever=db.as_retriever()
retriever.invoke(query)
docs[0].page_content
```

# SimilaritySearchwithscore

```python
docs_and_score=db.similarity_search_with_score(query)
docs_and_score
```

```python
embeddings_vector=embeddings.embed_query(query)
embeddings_vector
```

```python
docs_score=db.similarity_search_by_vector(embeddings_vector)
docs_score
```

```python
### Savings and Loading the Vector Store
db.save_local("faiss_index")
```

```python
new_df=FAISS.load_local("faiss_index", embeddings,allow_dangerous_deserialization=True)
docs=new_df.similarity_search(query)
docs[0].page_content
```
