```python
import os
from dotenv import load_dotenv
load_dotenv

os.environ['OPENAI_API_KEY'] = os.getenv('OPENAI_API_KEY')
os.environ['LANGCHAIN_API_KEY'] = os.getenv('LANGCHAIN_API_KEY')   #langsmith tracking
os.environ["LANGCHAIN_TRACING_V2"] = "true"  #langsmith tracing
os.environ["LANGCHAIN_PROJECT"] = os.getenv('LANGCHAIN_PROJECT')  #langsmith project name
```

```python
from langchain_openai import ChatOpenAI
llm=ChatOpenAI(model="gpt-4o")
print(llm)
```

```python
## Input and get response from llm

result=llm.invoke("What is Gen AI?")
```

```python
print(result)
```

```python
## Chatpromt Template
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(
    [
        ("System", "You are expert AI engineer. Provide me answers based on questions")
        ("user", "{input}")
    ]
)
```

```python
chain=prompt|llm

response=chain.invoke({"input":"Can you tell me about langsmith?"})
print(response)
```

```python
type(response)
```

```python
## stroutput Parser
from langchain_core.output_parsers import StrOutputParser
output_parser=StrOutputParser
chain=prompt|llm|output_parser

response=chain.invoke({"input":"Can you tell me about the langsmith?"})
print(response)
```
